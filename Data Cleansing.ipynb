### Analizando la columna **PoolQC**

Según **data_description.txt**


PoolQC: Pool quality

       Ex	Excellent
       Gd	Good
       TA	Average/Typical
       Fa	Fair
       NA	No Pool
       
Borramos información vital en el DataFrame.

En lugar de borrarla la podemos convertir en  una columna de datos  ordinales.

```python
def descripcion_variable(nombre_variable, ruta="house-prices-advanced-regression-techniques/data_description.txt"):
    with open(ruta, "r", encoding="utf-8", errors="ignore") as f: #esto me permite no modificar el archivo original: lo copia en f y desde ahi trabajo
        lineas = f.readlines()

    resultado = []
    capturando = False

    for linea in lineas:
        if linea.strip().startswith(nombre_variable + ":"): #el strip salta espacios. 
            capturando = True

        if capturando:          
            if linea.strip() != "" and ":" in linea and not linea.strip().startswith(nombre_variable + ":"): #examina la siguiente linea, y si no es el nombre de la variable, fuera del puto bucle
                break
            resultado.append(linea.rstrip())

    return "\n".join(resultado)



print(descripcion_variable("PoolQC"))
```

# Cuartiles y  datos atipicos

### Limpieza por cuartiles

Los cuartiles dividen los datos en cuatro partes iguales: Q1 corresponde al 25% inferior, Q2 es la mediana y Q3 al 75% inferior. En un **boxplot** la caja se extiende desde Q1 hasta Q3, con una línea en la mediana (Q2). Los bigotes se alargan hasta $1.5$ veces el rango intercuartílico ($IQR = Q3 − Q1$) a partir de los bordes de la caja, y los valores que quedan fuera de este rango se consideran atípicos.


```python
%matplotlib inline
df.boxplot('SalePrice')
```

```python
sP = df['SalePrice']
IQR = 1.5*(sP.quantile(.75)-sP.quantile(.25))
lim_sup = sP.quantile(.75)+IQR
lim_inf = sP.quantile(.25)-IQR
```

```python
sP_clean = sP[(sP >= lim_sup) | (sP <= lim_inf)] #| es la funcion lógica or
sP_clean
```

```python
df.drop(sP_clean.index)
```

### Limpieza por desviación estandar

La desviación estándar es una medida estadística que indica el grado de dispersión de los datos respecto a la media. Un valor alto refleja datos muy dispersos, mientras que un valor bajo señala que los datos están más concentrados alrededor de la media. El puntaje Z (**Z-score**) expresa cuántas desviaciones estándar se encuentra una observación por encima o por debajo de la media de la distribución.

```python
from scipy import stats
sP = df['SalePrice']
z = np.abs(stats.zscore(sP))
sP_clean = sP[(z>3) | (z<-3)]
df.drop(sP_clean.index)
```
